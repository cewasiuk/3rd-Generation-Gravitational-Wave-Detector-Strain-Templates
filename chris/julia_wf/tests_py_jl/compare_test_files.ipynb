# Python vs Julia output comparison

Run the Python and Julia generators first, then compare the text files they write into `test_files/python` and `test_files/julia`.

```python
from pathlib import Path
import numpy as np

BASE = Path.cwd()
PY_DIR = BASE / 'test_files' / 'python'
JL_DIR = BASE / 'test_files' / 'julia'

def load_file(path):
    data = np.loadtxt(path)
    data = np.asarray(data)
    if data.ndim == 0:
        return data.reshape(1)
    if data.ndim == 1:
        return data
    if data.ndim == 2 and data.shape[1] == 2:
        return data[:, 0] + 1j * data[:, 1]
    return data.reshape(-1)

def compare_arrays(py, jl):
    py = np.asarray(py).reshape(-1)
    jl = np.asarray(jl).reshape(-1)
    shared = min(py.size, jl.size)
    if shared == 0:
        return 0.0, 0.0, shared
    delta = py[:shared] - jl[:shared]
    max_abs = float(np.max(np.abs(delta)))
    max_rel = float(np.max(np.abs(delta) / np.maximum(1.0, np.abs(py[:shared]))))
    return max_abs, max_rel, shared
```

```python
rows = []
for py_file in sorted(PY_DIR.glob('*.txt')):
    jl_file = JL_DIR / py_file.name
    if not jl_file.exists():
        rows.append((py_file.name, 'missing', None, None, None, None))
        continue
    py = load_file(py_file)
    jl = load_file(jl_file)
    max_abs, max_rel, shared = compare_arrays(py, jl)
    rows.append((py_file.name, 'ok', max_abs, max_rel, py.size, jl.size, shared))

for row in rows:
    if row[1] == 'missing':
        print(f'{row[0]}: missing counterpart')
    else:
        name, _, max_abs, max_rel, py_size, jl_size, shared = row
        print(f'{name}: max_abs={max_abs:.3e} max_rel={max_rel:.3e} py={py_size} jl={jl_size} shared={shared}')
        if max_rel > 1e-10:
            print(f'  -> mismatch detected: max_abs={max_abs:.3e} max_rel={max_rel:.3e}')

mismatched = [row for row in rows if row[1] == 'ok' and row[4] != row[5]]
print('shape_mismatches=', [row[0] for row in mismatched])
```
