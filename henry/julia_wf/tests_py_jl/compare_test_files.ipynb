# Python vs Julia Henry Waveform Checks

Run the Python and Julia generators first, then execute the comparison cell.

```python
from pathlib import Path
import numpy as np
import pandas as pd

base = Path('test_files')
py_dir = base / 'python'
jl_dir = base / 'julia'

rows = []
for py_path in sorted(py_dir.glob('*.txt')):
    jl_path = jl_dir / py_path.name
    if not jl_path.exists():
        rows.append({'file': py_path.name, 'status': 'missing julia'})
        continue
    py = np.loadtxt(py_path, ndmin=2)
    jl = np.loadtxt(jl_path, ndmin=2)
    if py.shape != jl.shape:
        rows.append({'file': py_path.name, 'status': f'shape {py.shape} vs {jl.shape}'})
        continue
    diff = np.abs(py - jl)
    denom = np.maximum(np.abs(py), 1e-300)
    rows.append({
        'file': py_path.name,
        'status': 'ok',
        'max_abs': diff.max(),
        'max_rel': (diff / denom).max(),
    })

pd.DataFrame(rows).sort_values('file')
```
